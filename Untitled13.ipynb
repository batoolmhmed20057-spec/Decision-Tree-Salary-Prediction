```python
import pandas as pd
import numpy as np 
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeRegressor,plot_tree
df = pd.read_csv("ai pioneres dataset")
print("Dataset loaded successfully! shape:",df.shape)
```

```python
# check column names and data types
print(df.columns.tolist())
#check missing values
print(df.isnull().sum())
#Display salary examples
print(df["salary"].head(10))
```

```python
# define input features
x = df[["company" ,"rating" ,"location" ,"positionName"]]
#define target variable
y = df["salary"]
print("Features:")
display(x.head())
print("target:")
display(y.head())
```

```python
df["rating"] = df["rating"].fillna(df["rating"].median())
print("Missing rating:",df["rating"].isnull().sum())
```

```python
x =pd.get_dummies(
    df[["company" ,"rating" ,"location" ,"positionName"]],
    columns=["company" , "location" , "positionName"],
    dtype=int
)

display(x.head())
print("Encoded shape:",x.shape)
```

```python
from sklearn.tree import DecisionTreeRegressor
import pandas as pd

x_clean = x.fillna(0)


y_numeric = pd.to_numeric(df["salary"], errors='coerce').fillna(0)


model = DecisionTreeRegressor(random_state=42)
model.fit(x_clean, y_numeric)

print("Model trained successfully!")
```

```python
y_pred = model.predict(x)

comparison = pd.DataFrame({
    "Actual Salary":y,
    "predicted salary":y_pred
})
display(comparison.head(10))
```

```python
import re
from sklearn.tree import DecisionTreeRegressor, plot_tree
import matplotlib.pyplot as plt

# 1. استخراج الأرقام من عمود salary وتنظيفه من أي رموز أو نصوص
def extract_number(val):
    if pd.isna(val):
        return None
    # البحث عن الأرقام والنقاط فقط
    nums = re.findall(r'[\d\.]+', str(val).replace(',', ''))
    if nums:
        return float(nums[0])
    return None

# تطبيق التنظيف على عمود salary
df['clean_salary'] = df['salary'].apply(extract_number)

# 2. ملء القيم الفارغة بمتوسط المرتبات (وليس صفر)
mean_salary = df['clean_salary'].mean()
y_numeric = df['clean_salary'].fillna(mean_salary)

# 3. تجهيز x وتدريب النموذج
x_clean = x.fillna(0)

model = DecisionTreeRegressor(max_depth=3, random_state=42)
model.fit(x_clean, y_numeric)

# 4. رسم الشجرة
plt.figure(figsize=(20, 10))
plot_tree(
    model,
    feature_names=x_clean.columns,
    filled=True,
    rounded=True,
    fontsize=10
)
plt.title("Decision Tree Regressor")
plt.show()
```
